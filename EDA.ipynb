```python
# ============================================================
# PRAKTIKUM 1 - OPERASI DASAR NUMPY ARRAY
# ============================================================

import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])

print("Rata-rata harga:", harga.mean())
print("Harga tertinggi:", harga.max())
print("Harga setelah diskon 10%:", harga * 0.9)


# ============================================================
# PRAKTIKUM 2 - MEMBUAT SERIES & DATAFRAME
# ============================================================

import pandas as pd

data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}

df = pd.DataFrame(data_kantin)

print("\n===== DATA KANTIN =====")
print(df)


# ============================================================
# PRAKTIKUM 3 - DATA LOADING & INSPECTION
# ============================================================

data_kantin = {
    'menu': [
        'Nasi Goreng',
        'Es Teh',
        'Mie Ayam',
        'Es Teh',
        None,
        'Bakso',
        'Nasi Goreng'
    ],
    'harga': [
        12000,
        4000,
        10000,
        4000,
        8000,
        9000,
        12000
    ],
    'terjual': [
        23,
        40,
        None,
        35,
        18,
        25,
        23
    ]
}

data = pd.DataFrame(data_kantin)

data.to_csv('data_kantin.csv', index=False)

df = pd.read_csv('data_kantin.csv')

print("\n===== 5 BARIS PERTAMA =====")
print(df.head())

print("\n===== INFORMASI DATA =====")
df.info()

print("\n===== STATISTIK DATA =====")
print(df.describe())

print("\n===== UKURAN DATA =====")
print(df.shape)


# ============================================================
# PRAKTIKUM 4 - MENANGANI MISSING VALUE
# ============================================================

print("\n===== JUMLAH MISSING VALUE =====")
print(df.isnull().sum())

df['terjual'] = df['terjual'].fillna(0)

df = df.dropna(subset=['menu'])

print("\n===== DATA SETELAH CLEANING =====")
print(df)

print("\n===== CEK MISSING VALUE =====")
print(df.isnull().sum())


# ============================================================
# PRAKTIKUM 5 - DUPLIKAT DAN TIPE DATA
# ============================================================

print("\n===== JUMLAH DATA DUPLIKAT =====")
print(df.duplicated().sum())

df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int)

print("\n===== DATA SETELAH DROP DUPLICATES =====")
print(df)

print("\n===== TIPE DATA =====")
print(df.dtypes)


# ============================================================
# PRAKTIKUM 6 - DATA MANIPULATION
# ============================================================

# Filtering
laris = df[df['terjual'] > 20]

print("\n===== MENU TERJUAL LEBIH DARI 20 =====")
print(laris)

# Sorting
urut = df.sort_values(by='terjual', ascending=False)

print("\n===== DATA BERDASARKAN PENJUALAN =====")
print(urut)

# Kolom turunan
df['total_pendapatan'] = df['harga'] * df['terjual']

print("\n===== DATA DENGAN TOTAL PENDAPATAN =====")
print(df)

# Groupby
ringkasan = df.groupby('menu')['total_pendapatan'].sum()

print("\n===== TOTAL PENDAPATAN SETIAP MENU =====")
print(ringkasan)

# Pendapatan tertinggi
menu_tertinggi = ringkasan.idxmax()
pendapatan_tertinggi = ringkasan.max()

print("\n===== PENDAPATAN TERTINGGI =====")
print("Menu:", menu_tertinggi)
print("Total pendapatan:", pendapatan_tertinggi)
```
